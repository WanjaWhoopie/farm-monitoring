# Behaviour experiments

A small behaviour model should hold up on a day it has not seen, and the pen activity budget should stay close to the true share of time. The single-date crop score is the earlier check. The date 2019-12-02 is never used for training or for choosing a model.

Scripts are in `../scripts/`. Each one is the procedure that wrote the table for that experiment. From that folder, pass `--data` as the annotated clips (`output.json`, `times.txt`, `color.mp4`). The command writes `history.csv` and `validation_report.csv`. Measured tables already saved are in `../results/`. The code cell below lists those result folders.

| Notebook | What it records | Verdict |
| --- | --- | --- |
| `01_frozen_backbone.ipynb` | MobileNet head only, one validation date | keep as the model-only reference |
| `02_full_finetune.ipynb` | Unfrozen backbone at 1e-4 | do not retry |
| `03_differential_lr.ipynb` | Backbone 1e-5, head 1e-4 | do not retry |
| `04_temporal_window.ipynb` | Mean of five crops, frozen head | keep: best crop result |
| `05_feature_pool.ipynb` | Average of five frozen embeddings | do not retry |
| `06_geometry_one_day.ipynb` | Nine box features, one validation date | keep on that date only |
| `07_farm_protocol.ipynb` | Leave-one-day-out budget, sequence, and fusion | keep the box model; do not retry the other two |

The budget comparison is `07_farm_protocol.ipynb`. The crop notebooks are why an image network was not kept as the whole model.


```python
from pathlib import Path
import pandas as pd

def results(name):
    here = Path.cwd()
    candidates = [
        here / "projects" / "pig-video" / "results" / name,
        here.parent / "results" / name,
        here / "results" / name,
    ]
    for path in candidates:
        if path.is_dir():
            return path
    raise FileNotFoundError(name)
print('Result folders:')
for path in sorted((results('e1-frozen-backbone').parent).iterdir()):
    print('-', path.name)
```
