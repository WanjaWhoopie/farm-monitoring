```python
from pathlib import Path
import pandas as pd

def results(name):
    here = Path.cwd()
    candidates = [
        here / "projects" / "pig-video" / "results" / name,
        here.parent / "results" / name,
        here / "results" / name,
    ]
    for path in candidates:
        if path.is_dir():
            return path
    raise FileNotFoundError(name)
folder = results('g1-geometry-hgb')
display(pd.read_csv(folder / 'history.csv'))
display(pd.read_csv(folder / 'validation_report.csv', index_col=0))
```
